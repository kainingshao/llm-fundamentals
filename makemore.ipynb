```python
import torch
```

```python
words = open('names.txt', 'r').read().splitlines()
N = torch.zeros((28, 28), dtype=torch.int32)
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i + 1 for i, s in enumerate(chars)}
stoi['.'] = 0
index_to_char = {i:s for s, i in stoi.items()}
for word in words:
    chars = ['.'] + list(word) + ['.']
    for char1, char2 in zip(chars, chars[1:]):
        char1, char2
        index1 = stoi[char1]
        index2 = stoi[char2]
        N[index1, index2] += 1
```

```python

import matplotlib.pyplot as pyplot
%matplotlib inline

pyplot.figure(figsize=(16, 16))
pyplot.imshow(N, cmap='Blues')
for i in range(27):
    for j in range(27):
        bigram = index_to_char[i] + index_to_char[j]
        pyplot.text(j, i, bigram, ha='center', va='bottom', color = 'gray')
        pyplot.text(j, i, N[i ,j].item(), ha='center', va='top', color = 'gray')
pyplot.axis('off')
```
